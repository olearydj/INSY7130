# Data-work bridge for 01b

## A short refresher, not new course content

Use this optional notebook before 01b if you would like a quick practice with
four pandas moves that the main notebook assumes:

1. select columns;
2. filter rows;
3. summarize groups; and
4. make a simple plot.

It uses the same UCI Wholesale customers data as 01b. The aim is to make the
later analytical conversation accessible, not to teach pandas comprehensively.

## 1. Load the file

The main notebook explains the data source and its limits. Here we only load
the file so that we can practice the four basic moves.

````python
import matplotlib.pyplot as plt
import pandas as pd
````

````python
DATA_URL = "https://archive.ics.uci.edu/static/public/292/wholesale%2Bcustomers.zip"

customers = pd.read_csv(DATA_URL, compression="zip")
customers.head()
````

## 2. Select columns

Square brackets with a list select a small set of columns. This is useful when
you want to examine only the fields needed for a question.

````python
customers[["Channel", "Region", "Grocery"]].head()
````

Change the list to inspect one or two other spending columns. The source uses
numbers for `Channel`: 1 is Horeca and 2 is Retail.

## 3. Filter rows

A Boolean condition selects the rows for which the condition is true. This
cell keeps only the rows whose existing channel code is 2.

````python
retail_customers = customers[customers["Channel"] == 2]

retail_customers[["Channel", "Grocery"]].head()
````

Use `shape` to compare the number of Retail records with the number in the
whole file.

````python
customers.shape, retail_customers.shape
````

## 4. Summarize groups

`groupby` splits the rows into groups; `agg` calculates one or more summaries
for each group. This is the main pandas pattern used in 01b.

````python
customers.groupby("Channel")["Grocery"].agg(["count", "mean", "median"])
````

Read the result one row at a time. It gives the number of records, mean annual
grocery spending, and median annual grocery spending for each existing channel.

## 5. Make a simple plot

A grouped box plot lets us see more of each group's spending distribution than
a single mean or median. The main notebook uses the same kind of plot to
interpret its declared statistic.

````python
fig, ax = plt.subplots(figsize=(7, 5))

customers.boxplot(column="Grocery", by="Channel", ax=ax)
ax.set_title("Annual grocery spending by existing channel")
ax.set_xlabel("Existing channel code: 1 = Horeca, 2 = Retail")
ax.set_ylabel("Annual grocery spending (source monetary units)")
plt.suptitle("")
plt.show()
````

## Ready for 01b

You are ready to continue when you can recognize these four patterns:

```text
customers[["column_a", "column_b"]]        # select columns
customers[customers["column_a"] == value]  # filter rows
customers.groupby("group")["value"].agg(...)  # summarize groups
customers.boxplot(...)                       # make a simple plot
```

In 01b, the important work is not memorizing these forms. It is using them to
decide what the file represents, make a transparent descriptive comparison, and
state the limits of the resulting claim.
